# E-Commerce Purchase Behavior Analysis

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
import warnings; warnings.filterwarnings('ignore')
%matplotlib inline
import utils
from google.colab import files

uploaded = files.upload()


df = utils.load_data(filepath="orders.csv")
df.head()
```

## Structure & Data Types

```python
print(df.dtypes.value_counts().to_dict())
df.info()
```

## Missing Values

```python
mv= utils.missing_report(df)
print('Columns with missing values:', len(mv))
print(mv.head(10).to_string())
```

## Numeric Distributions

```python
num= ['price', 'freight', 'payment', 'review_score']
num= [c for c in num if c in df.columns]
fig, ax= plt.subplots(1,len(num), figsize=(4*len(num),3))
ax= np.atleast_1d(ax)
for a, c in zip(ax, num):
  a.hist(df[c].dropna(), bins= 40, color= '#4c72b0'); a.set_title(c)
plt.tight_layout(); plt.show()
df[num].describe().round(2)
```

## Top Categories

```python
vc= utils.top_counts(df['payment_type'],10)
fig, ax= plt.subplots(figsize=(9,5)); vc.plot(kind= 'bar', ax=ax, color='#55a868'); ax.set_title('Top Payment Method'); plt.xticks(rotation= 45, ha= 'right'); plt.tight_layout(); plt.show()
print(vc.to_string())
```
